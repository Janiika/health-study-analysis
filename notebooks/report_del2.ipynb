# Medelblodtryck - 95% konfidensintervall (normalapproximation & bootstrap)

```python
systolic_bp = df["systolic_bp"].dropna()
n = len(systolic_bp)

# Konfidensintervall med normalapproximation
mean_bp = systolic_bp.mean()
sd = systolic_bp.std(ddof=1)
se = sd / np.sqrt(n)
z = 1.96
half_width = z * se
lo, hi = mean_bp - half_width, mean_bp + half_width

# Konfidensintervall med bootstrap
np.random.seed(123)

B = 1000
boot_means = []
for b in range(B):
    boot_sample = np.random.choice(systolic_bp, size=n, replace=True)
    boot_means.append(boot_sample.mean())

blo, bhi = np.percentile(boot_means, [2.5, 97.5])

print(f"""
---------- Medelblodtryck med konfidensintervall ----------
Antal observationer             : {n}
Observerad medel                : {mean_bp:.1f} mmHg
Standardavvikelse               : {sd:.3f}
Standarderror                   : {se:.3f}

95%-CI med normalapproximation  : ({lo:.2f}, {hi:.2f})
95%-CI med bootstrap            : ({blo:.2f}, {bhi:.2f})
""")
```

Konfidensintervall beskriver osäkerheten i skattningen av populationens sanna genomsnittliga värde. (Källa: lektion `6.2 - Konfidensintervall`)

- Det observerade medelvärdet för systolisk blodtryck var cirka 149,2 mmHg. Ett 95-procentigt konfidensintervall beräknades till cirka 148,3 - 150,1 mmHg.
- Båda metoderna (normalapproximation och bootstrap) gav i princip samma konfidensintervall. Detta tyder på att normalapproximationen fungerar bra på detta dataset, sannolikt eftersom stickprovet är stort (n = 800) och relativt normalfördelad utan tydlig skevhet.

# Hypotesprövning - har **rökare** högre blodtryck än **icke-rökare**?

**Hypoteser**
- H0: Medelblodtrycket rökare = icke-rökare
- H1: Medelblodtrycket rökare > icke-rökare

```python
from scipy.stats import ttest_ind

smokers = df.loc[df["smoker"] == "Yes", "systolic_bp"].dropna()
non_smokers = df.loc[df["smoker"] == "No", "systolic_bp"].dropna()

n_smokers = len(smokers)
n_non_smokers = len(non_smokers)

alpha = 0.05

mean_smokers = smokers.mean()
mean_non_smokers = non_smokers.mean()

# Ensidig t-test: rökare > icke-rökare (medelblodtryck), Welch-variant
t_stat, p_value = ttest_ind(
    smokers,
    non_smokers,
    equal_var=False,
    alternative="greater"
)

# Tvåsidigt t-test för att få fram ett konfidensintervall
result_two_sided = ttest_ind(
    smokers,
    non_smokers,
    equal_var=False,
)

ci = result_two_sided.confidence_interval(confidence_level=0.95)

print(f"""
----------- Medelblodtryck rökare vs. icke-rökare -----------
Antal rökare                : {n_smokers}
Antal icke-rökare           : {n_non_smokers}
Medelblodtryck rökare       : {mean_smokers:.1f} mmHg
Medelblodtryck icke-rökare  : {mean_non_smokers:.1f} mmHg
Observerad skillnad         : {(mean_smokers - mean_non_smokers):.1f} mmHg

t-stat                      : {t_stat}
p-värde (ensidigt)          : {p_value}
95%-CI för skillnaden       : ({ci.low:.2f}, {ci.high:.2f}) mmHg
""")

if p_value < alpha:
    print(f"Nollhypotesen förkastas (p = {p_value} < {alpha}).")
else:
    print(f"Nollhypotesen kan inte förkastas (p = {p_value:.4f} > {alpha}).")
```

Welch-test användes eftersom grupperna är oberoende och eftersom metoden inte kräver att grupperna har samma varians. Testet passar för jämförelser av medelvärden när den undersökta variabeln är kontinuerlig. Eftersom hypotesen var riktad (rökare > icke-rökare) användes ett ensidigt test. Tvåsidigt test användes för att få fram ett konfidensintervall. (Källa: lektion `7.3 - A/B-test från princip till praktik`)
- Rökare hade något högre observerat genomsnittligt systoliskt blodtryck än icke-rökare (cirka 149,5 vs. 149,1 mmHg), men skillnaden var liten, endast cirka 0,5 mmHg.
- Konfidensintervallet innehåller noll, och p-värdet är betydligt högre än signifikansnivån 0.05. Nollhypotesen kan därför inte förkastas.
- Resultatet ger därför inte statistiskt stöd för hypotesen att rökare har högre medelblodtryck än icke-rökare i detta dataset.

# Simulering power - hur säker är hypotestestet?

```python
np.random.seed(5)

# Simulerad poweranalys: genererar nya stickprov baserade på de observerade värden och
# beräknar hur ofta Welch-testet upptäcker den observerade skillnaden (p < 0.05).
n_simulations = 1000
alpha = 0.05

n_smokers = len(smokers)
n_non_smokers = len(non_smokers)

sd_smokers = smokers.std(ddof=1)
sd_non_smokers = non_smokers.std(ddof=1)

mean_smokers = smokers.mean()
mean_non_smokers = non_smokers.mean()

significant_results = 0

for _ in range(n_simulations):
    sim_smokers = np.random.normal(
        loc=mean_smokers,
        scale=sd_smokers,
        size=n_smokers
    )

    sim_non_smokers = np.random.normal(
        loc=mean_non_smokers,
        scale=sd_non_smokers,
        size=n_non_smokers
    )

    t_stat, p_value = ttest_ind(
        sim_smokers,
        sim_non_smokers,
        equal_var=False
    )

    if p_value < alpha:
        significant_results += 1

power = significant_results / n_simulations

print(f"""
---------- Hypotestestets power ----------
Antal simuleringar          : {n_simulations}
Andel signifikanta resultat : {power:.3f}
Uppskattad power            : {power:.1%}
""")
```

Powern uppskattades genom simulering där nya stickprov genererades utifrån de observerade medelvärdena och standardavvikelserna. Metoden gör det möjligt att undersöka hur ofta testet upptäcker en verklig skillnad av den storlek som observerades i datasetet (cirka 0,5 mmHg).
- Den uppskattade powern var cirka 6,4% -> detta innebär att Welch-testet upptäckte den observerade skillnaden som statistiskt signifikat i ungefär 6 av 100 simuleringar. Powern är därmed betydligt lägre än den ofta använda målnivån på 80%.
- Den låga powern visar att det är svårt att upptäcka en skillnad av denna storlek med den aktuella variationen och stickprovsstorleken.

# Stickprovsstorlek för 80% power

```python
from statsmodels.stats.power import TTestIndPower

analysis = TTestIndPower()

effect = mean_smokers - mean_non_smokers
power = 0.80
alpha = 0.05

pooled_sd = np.sqrt(
    (
        sd_smokers**2 +
        sd_non_smokers**2
    ) / 2
)

effect_size = effect / pooled_sd

required_n = analysis.solve_power(
    effect_size=effect_size,
    power=power,
    alpha=alpha,
    ratio=1
)

print(f"""
---------- Stickprovsstorlek för 80% power ----------
Observerad skillnad                     : {effect:.1f} mmHg
Target power                            : {power * 100}%
Effect size (Cohen's d)                 : {effect_size:.4f}
Nödvändig stickprovsstorlek per grupp   : {np.ceil(required_n):.0f}
""")
```

Power beskriver sannolikheten att ett hypotestest hittar en verklig skillnad när den faktiskt finns. För att undersöka hur stor stickprovsstorlek som skulle krävas för att upptäcka den observerade skillnaden med 80% power beräknades Cohens d och analyserades med hjälp av TTestIndPower från statsmodels.
- Effektstorleken (Cohen's d) uppskattades till 0,036 vilket motsvarar en mycket liten effekt (< 0,2).
- Beräkningen visar att cirka 11 825 individer per grupp skulle krävas för att upptäcka en skillnad av denna storlek på ett tillförlitligt sätt.

Källor: lektion `6.3 - Hypotesprövning` och `7.4 - Power och flerfaldig testning`

# Multipel regression - blodtryck

```python
df_copy = df.copy()

df_copy["bmi"] = su.calculate_bmi(df_copy)
df_copy["smoker_yes"] = (df_copy["smoker"] == "Yes").astype(int)
df_copy["sex_male"] = (df_copy["sex"] == "M").astype(int)

X = df_copy[["age", "bmi", "cholesterol", "disease", "smoker_yes", "sex_male"]]
y = df_copy["systolic_bp"]

linreg = LinearRegression()
linreg.fit(X, y)

intercept = float(linreg.intercept_)
r2 = float(linreg.score(X, y))

print(f"""
----- Regression -----
Intercept           : {intercept:.2f}
R²                  : {r2:.3f}
""")

print("---- Koefficienter ----")
for variable, coef in zip(X.columns, linreg.coef_):
    print(f"{variable:<15}       : {coef:.3f}")
```

- En multipel linjär regression användes för att undersöka hur ålder, BMI, kolesterol, sjukdom, rökning och kön samvarierar med systoliskt blodtryck. Multipel regression är lämplig när flera prediktorer analyseras samtidigt eftersom varje regressionskoefficient kan tolkas som effekten av en variabel när övriga variabler hålls konstanta. Variablerna sjukdom, rökare och man är så kallade dummyvariabler som kodats som 1 = ja och 0 = nej.

- Den skattade multipla regressionsmodellen kan skrivas som:
**Systoliskt blodtryck = 111,07 + 0,52 * ålder + 0,34 * BMI + 0,55 * kolesterol + 0,57 * sjukdom - 0,31 * rökare + 2,40 * man**

- **Tolkning:** Modellen förklarade cirka **39,8% av variationen i blodtrycket** (R² ≈ 0,398). Samtliga variabler förutom rökning hade positiva regressionskoefficienter. Resultaten tyder på att högre ålder, BMI, kolesterolnivå, förekomst av sjukdom och manligt kön är förknippade med högre systoliskt blodtryck i datasetet. Exempelvis var ett års högre ålder associerat med cirka 0,52 mmHg högre blodtryck, medan koefficienten för kön var cirka 2,40 vilket innebär att **män** i genomsnitt förväntas ha **2,4 mmHg högre** systoliskt blodtryck än kvinnor när övriga variabler i modellen hålls konstanta.

- **OBS!** Eftersom variablerna mäts på olika skalor bör koefficienternas storlek inte jämföras direkt. En större koefficient innebär inte nödvändigtvis att variabeln har större betydelse för blodtrycket. R²-värdet beskriver hur stor andel av variationen i blodtrycket som förklaras av modellen som helhet, inte hur viktig en enskild variabel är.

Källa: https://casrai.org/guides/how-to-interpret-r-squared

I den här regressionsmodellen fick variablen rökning en liten negativ koefficient (-0,31), medan den observerade jämförelsen mellan grupperna på hypotestestet visade ett något högre medelblodtryck bland rökare (+0,5 mmHg). Detta är inte nödvändigtvis motsägelsefullt eftersom regressionsmodellen samtidigt kontrollerar för ålder, BMI, kolesterol, sjukdom och kön. Koefficienten för rökning beskriver därför sambandet när övriga variabler hålls konstansta, medan t-testet endast jämför gruppernas observerade medelevärden.  

En förenklat exempel är att rökare i genomsnitt kan vara äldre än icke-rökare i datasetet. Eftersom högre ålder är förknippad med högre blodtryck kan rökare då få ett högre observerat medelblodtryck, även om själva rökningen inte har någon tydlig effekt. När regressionsmodellen tar hänsyn till ålder och andra faktorer kan sambandet därför förändras eller till och med byta riktning.
